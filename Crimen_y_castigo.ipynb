```python
from google.colab import files

archivo = files.upload()

nombre_archivo = list(archivo.keys())[0]

with open(nombre_archivo, "r", encoding="utf-8") as documento:
    texto = documento.read()

print(texto[:1000])
```

```python
texto = texto.lower()

print(texto[:1000])
```

```python
import re

texto_limpio = re.sub(r'[^a-záéíóúüñ\s]', ' ', texto)

print(texto_limpio[:1000])
```

```python
palabras = texto_limpio.split()

print(palabras[:50])
```

```python

stopwords = {
    "el", "la", "los", "las", "un", "una", "unos", "unas",
    "de", "del", "al", "a", "ante", "bajo", "con", "contra",
    "desde", "en", "entre", "hacia", "hasta", "para", "por",
    "según", "sin", "sobre", "tras",
    "y", "e", "ni", "o", "u", "pero", "sino", "aunque",
    "que", "quien", "cual", "cuales", "cuyo",
    "se", "me", "te", "nos", "lo", "le", "les",
    "mi", "mis", "tu", "tus", "su", "sus",
    "yo", "tú", "él", "ella", "ellos", "ellas",
    "es", "son", "era", "eran", "ser", "estar",
    "estaba", "estaban", "ha", "han", "había",
    "he", "hay", "fue", "fueron", "muy", "más",
    "menos", "ya", "no", "sí", "como", "cuando",
    "donde", "también", "todo", "toda", "todos",
    "todas", "este", "esta", "estos", "estas",
    "aquel", "aquella", "aquellos", "aquellas",
    "don", "doña", "así", "pues", "será"
}
```

```python
palabras_filtradas = [
    palabra for palabra in palabras
    if palabra not in stopwords and len(palabra) > 2
]

print(palabras_filtradas[:50])

print("Total de palabras originales:", len(palabras))
print("Total de palabras filtradas:", len(palabras_filtradas))
```

```python
from collections import Counter

frecuencias = Counter(palabras_filtradas)

print("Las 20 palabras más frecuentes:")

for palabra, cantidad in frecuencias.most_common(20):
    print(palabra, ":", cantidad)
```

```python

import matplotlib.pyplot as plt
from wordcloud import WordCloud

nube = WordCloud(
    width=1200,
    height=700,
    background_color="white",
    colormap="viridis",
    max_words=100,
    collocations=False
).generate_from_frequencies(frecuencias)

plt.figure(figsize=(14, 8))
plt.imshow(nube, interpolation="bilinear")
plt.axis("off")
plt.title("Nube de palabras - Crimen y Castigo",
          fontsize=18)

plt.show()
```

```python
nube.to_file("wordcloud_crimen_y_castigo.png")

files.download("wordcloud_crimen_y_castigo.png")
```

Para el procesamiento del primer capítulo del libro de Crimen y Castigo, decidí convertir todas las palabras a minúsculas con la finalidad de unificar su escritura y evitar que Python contabilice como términos diferentes a las palabras que solo se distinguen por el uso de mayúsculas. Asimismo, eliminé los signos de puntuación, los números y los caracteres especiales, debido a que no aportan información relevante para identificar las palabras más frecuentes. Sin embargo, se conservaron las letras con tilde y la letra ñ para mantener la escritura correcta del español.

Respecto a las stopwords, se seleccionaron artículos, preposiciones, conjunciones, pronombres y otras palabras de uso frecuente que no aportan un significado específico al análisis. Su eliminación permitió destacar los términos con mayor contenido informativo. Estas decisiones influyen directamente en las frecuencias obtenidas, ya que reducen la presencia de palabras gramaticales y permiten identificar los conceptos que predominan en el capítulo. Finalmente, la nube de palabras representa visualmente estos resultados mediante diferentes tamaños, de acuerdo con la frecuencia de aparición de cada término.
